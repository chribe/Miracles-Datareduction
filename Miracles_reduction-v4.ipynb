# Some preliminary analysis

fit elastic with a gaussian and spectrum with a corresponding voigt to see if we match the initial values

```python
import lmfit as lm
import numpy as np
import scipy as sp
def gauss(x,w,s):
    return s/np.sqrt(2*np.pi*w**2)*np.exp(-x**2/2/w**2)
def voigt(x,sigma,w,scaling,ela):
    return scaling*sp.special.voigt_profile(x,sigma,w)+gauss(x,sigma,ela)
gaussmodel=lm.Model(gauss)
voigtmodel=lm.Model(voigt)
gaussparams=lm.Parameters()
gaussparams.add('w',min=0,value=10)
gaussparams.add('s',min=0,value=1)
voigtparams=lm.Parameters()
voigtparams.add('sigma',min=0,value=1,vary=False)
voigtparams.add('w',min=0,value=1)
voigtparams.add('scaling',min=0,value=1)
voigtparams.add('ela',min=0,value=1)
elastic=Result['data'].hist().transpose()
qens=Result1['data'].hist().transpose()
q=elastic.coords['qval'].values
q = 0.5 * (q[:-1] + q[1:])
hw=elastic.coords['hw'].values
hw=0.5 * (hw[:-1] + hw[1:])
gaussfits=[]
voigtfits=[]
widths=[]
dwidths=[]
for hiq,qv in enumerate(q):
    gaussfit=gaussmodel.fit(elastic['qval',hiq].values,x=hw,params=gaussparams)
    plt.figure()
    plt.plot(hw,elastic['qval',hiq].values)
    plt.plot(hw,gaussfit.eval())
    plt.plot(hw,gaussfit.init_fit)
    plt.title(r'$\sigma$='+f'{gaussfit.params['w'].value} ueV; FWHM={2*2**0.5*np.log(2)*gaussfit.params['w'].value} ueV')
    plt.yscale('log')
    gaussfits.append(gaussfit.eval())
    voigtparams['sigma'].value=gaussfit.params['w'].value
    plt.show()
    voigtfit=voigtmodel.fit(qens['qval',hiq].values,x=hw,params=voigtparams)
    voigtfits.append(voigtfit.eval())
    widths.append(voigtfit.params['w'].value)
    dw=voigtfit.params['w'].stderr
    if dw==None:dw=np.nan
    dwidths.append(dw)
vf=sc.DataArray(data=sc.array(dims=['qval','hw'],values=voigtfits,unit='counts'), coords={'qval': elastic.coords['qval'], 'hw':elastic.coords['hw']})
```

```python
pp.slicer({'elastic':elastic,'QENS':qens,'QENS Fit': vf},logy=True,ymin=1e-6,ymax=0.1,title=f'Energy axis shifted by {Result['treatment'].split('"')[1]}'+r' ueV',xlabel=r'$\hbar\omega$ [$\upmu$eV]')
```

```python
plt.figure()
plt.errorbar(q,widths,dwidths)
plt.plot([0.2,2],[10,10])
plt.xlabel(r'$q$ [\AA$^{-1}$]')
plt.ylabel(r'$\gamma$ [$\upmu$eV]')
plt.show()
```
